```python
import os
import pandas as pd
import streamlit as st

# Configuração inicial da página
st.set_page_config(page_title="Análise de Itens", layout="wide")
st.title("📊 Análise de Preços e Fretes (Olist)")

# Caminho dinâmico para evitar erros de arquivo não encontrado
diretorio_atual = os.path.dirname(os.path.abspath(__file__))
caminho_arquivo = os.path.join(diretorio_atual, "olist_order_items_dataset.csv")

# Leitura dos dados
tabela = pd.read_csv(caminho_arquivo, usecols=["price", "freight_value"])

# Contagens de valores
contagemprice = tabela["price"].value_counts().reset_index(name="Frequência")
contagemfreight = (
    tabela["freight_value"].value_counts().reset_index(name="Frequência")
)

# Organizando a tela do Streamlit em abas para ficar scaneável
aba1, aba2, aba3 = st.tabs(
    ["💰 Contagem de Preços", "🚚 Contagem de Fretes", "📋 Tabela Completa"]
)

with aba1:
    st.subheader("Frequência de Cada Preço")
    st.dataframe(contagemprice, use_container_width=True)

with aba2:
    st.subheader("Frequência de Cada Valor de Frete")
    st.dataframe(contagemfreight, use_container_width=True)

with aba3:
    st.subheader("Dados Brutos do Dataset")
    st.dataframe(tabela, use_container_width=True)
```

```python
import os
import pandas as pd
import streamlit as st

# Configuração da página do Streamlit
st.set_page_config(page_title="Análise de Pagamentos", layout="wide")
st.title("💳 Análise de Formas de Pagamento (Olist)")

# Caminho dinâmico para evitar erros de arquivo não encontrado
diretorio_atual = os.path.dirname(os.path.abspath(__file__))
caminho_arquivo = os.path.join(
    diretorio_atual, "olist_order_payments_dataset.csv"
)

# Leitura dos dados
tabela = pd.read_csv(caminho_arquivo, usecols=["payment_type", "payment_value"])

# Contagem dos tipos de pagamento formatada para tabela
contagempaymenttype = (
    tabela["payment_type"].value_counts().reset_index(name="Quantidade")
)

# Criando duas colunas lado a lado no Streamlit para os resultados
col1, col2 = st.columns(2)

with col1:
    st.subheader("Frequência por Tipo de Pagamento")
    st.dataframe(contagempaymenttype, use_container_width=True)

    st.subheader("Gráfico de Formas de Pagamento")
    # Cria um gráfico de barras automático usando os dados da contagem
    st.bar_chart(
        data=contagempaymenttype,
        x="payment_type",
        y="Quantidade",
        use_container_width=True,
    )

with col2:
    st.subheader("Dados Brutos das Transações")
    st.dataframe(tabela, use_container_width=True)
```

```python
import os
import pandas as pd
import streamlit as st

# Configuração da página do Streamlit
st.set_page_config(page_title="Análise de Produtos", layout="wide")
st.title("📦 Análise de Categorias e Fotos de Produtos (Olist)")

# Caminho dinâmico para evitar erros de arquivo não encontrado
diretorio_atual = os.path.dirname(os.path.abspath(__file__))
caminho_arquivo = os.path.join(diretorio_atual, "olist_products_dataset.csv")

# Leitura dos dados
tabela = pd.read_csv(
    caminho_arquivo, usecols=["product_category_name", "product_photos_qty"]
)

# Formatação das contagens para o Streamlit exibir corretamente
contagemcategory = (
    tabela["product_category_name"].value_counts().reset_index(name="Total")
)
contagemphotos = (
    tabela["product_photos_qty"].value_counts().reset_index(name="Total")
)

# Organização da tela em abas (Tabs)
aba1, aba2, aba3 = st.tabs(
    ["🏷️ Por Categoria", "📸 Por Qtd. de Fotos", "📋 Lista de Produtos"]
)

with aba1:
    st.subheader("Quantidade de Produtos por Categoria")
    st.dataframe(contagemcategory, use_container_width=True)

with aba2:
    st.subheader("Frequência de Produtos por Quantidade de Fotos")
    st.dataframe(contagemphotos, use_container_width=True)

with aba3:
    st.subheader("Dados Brutos do Dataset de Produtos")
    st.dataframe(tabela, use_container_width=True)
```

```python
import os
import pandas as pd
import streamlit as st

# Configuração da página do Streamlit
st.set_page_config(page_title="Clientes e Pagamentos", layout="wide")
st.title("👥 Análise de Clientes e Formas de Pagamento (Olist)")

# Caminho dinâmico para evitar erros de arquivo não encontrado
diretorio_atual = os.path.dirname(os.path.abspath(__file__))
caminho_customers = os.path.join(diretorio_atual, "olist_customers_dataset.csv")
caminho_payments = os.path.join(
    diretorio_atual, "olist_order_payments_dataset.csv"
)

# 1. LEITURA E CONTAGEM: DATASET CLIENTES
tabela_clientes = pd.read_csv(
    caminho_customers, usecols=["customer_city", "customer_state"]
)
contagem_cidades = (
    tabela_clientes["customer_city"].value_counts().reset_index(name="Clientes")
)
contagem_estados = (
    tabela_clientes["customer_state"]
    .value_counts()
    .reset_index(name="Clientes")
)

# 2. LEITURA E CONTAGEM: DATASET PAGAMENTOS
tabela_pagamentos = pd.read_csv(caminho_payments, usecols=["payment_type"])
contagem_pagamentos = (
    tabela_pagamentos["payment_type"].value_counts().reset_index(name="Total")
)

# ORGANIZAÇÃO NA TELA DO STREAMLIT (Abas)
aba1, aba2 = st.tabs(["📍 Visão Clientes", "💳 Visão Pagamentos"])

with aba1:
    st.subheader("Distribuição Geográfica dos Clientes")

    # Gráfico de barras interativo para os Estados
    st.bar_chart(
        data=contagem_estados,
        x="customer_state",
        y="Clientes",
        use_container_width=True,
    )

    col1, col2 = st.columns(2)
    with col1:
        st.markdown("**Clientes por Cidade**")
        st.dataframe(contagem_cidades, use_container_width=True)
    with col2:
        st.markdown("**Dados Brutos (Clientes)**")
        st.dataframe(tabela_clientes, use_container_width=True)

with aba2:
    st.subheader("Tipos de Pagamento Mais Utilizados")

    col3, col4 = st.columns(2)
    with col3:
        st.markdown("**Frequência de Pagamentos**")
        st.dataframe(contagem_pagamentos, use_container_width=True)
    with col4:
        st.markdown("**Dados Brutos (Pagamentos)**")
        st.dataframe(tabela_pagamentos, use_container_width=True)
```
