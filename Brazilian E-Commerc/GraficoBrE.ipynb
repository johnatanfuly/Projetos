# Gráficos de dados Brasileiros de E-Commerce da Olist 

Dataset Público de E-Commerce Brasileiro da Olist
Bem-vindo! Este é um dataset público de e-commerce brasileiro com pedidos feitos na Olist Store. O conjunto de dados possui informações de 100 mil pedidos realizados entre 2016 e 2018 em múltiplos marketplaces no Brasil. Seus recursos permitem visualizar um pedido sob diversas dimensões: desde o status do pedido, preço, pagamento e desempenho do frete até a localização do cliente, atributos do produto e, finalmente, avaliações escritas pelos consumidores. Também disponibilizamos um dataset de geolocalização que relaciona os CEPs brasileiros a coordenadas de latitude e longitude.
Estes são dados comerciais reais, foram anonimizados e as referências às empresas e parceiros nos textos de avaliação foram substituídas pelos nomes das grandes casas de Game of Thrones.

```python
import pandas as pd
import plotly.express as px

# 1. Carrega os dados otimizados
tabela = pd.read_csv("olist_customers_dataset.csv", usecols=["customer_state"])

# 2. Calcula a porcentagem mantendo os valores numéricos decimais (ex: 0.421)
contagem_state = (
    tabela["customer_state"]
    .value_counts(normalize=True)
    .reset_index(name="Porcentagem")
)

# 3. Cria o gráfico de barras com melhorias visuais e interativas
graph_state = px.bar(
    contagem_state,
    x="customer_state",
    y="Porcentagem",
    title="<b>Distribuição de Clientes por Estado</b>",
    labels={
        "customer_state": "Estado",
        "Porcentagem": "Proporção de Clientes",
    },
    text_auto=".1%",  # Mostra a porcentagem formatada no topo de cada barra automaticamente!
    color="Porcentagem",  # Adiciona um gradiente de cor baseado no tamanho da barra
    color_continuous_scale="Viridis",
)

# 4. Ajustes finos no layout
graph_state.update_layout(
    title_x=0.5,  # Centraliza o título do gráfico
    xaxis_tickangle=-45,  # Inclina os nomes dos estados para melhor leitura
    showlegend=False,  # Remove a legenda lateral repetitiva
)

# 5. Exibe o gráfico
graph_state.show()
```

```python
import pandas as pd
import plotly.express as px

# 1. Carrega os dados brutos (sem aplicar value_counts)
tabela = pd.read_csv(
    "olist_order_items_dataset.csv", usecols=["price", "freight_value"]
)

# 2. Remove valores discrepantes (outliers) para os gráficos ficarem legíveis
tabela_filtrada_preco = tabela[tabela["price"] <= 600]
tabela_filtrada_frete = tabela[tabela["freight_value"] <= 100]

# 3. Cria o histograma de Preços diretamente dos dados brutos
graph_price = px.histogram(
    tabela_filtrada_preco,
    x="price",
    title="<b>Distribuição dos Preços dos Produtos (Até R$ 600)</b>",
    labels={"price": "Preço (R$)"},
    color_discrete_sequence=["#1f77b4"],  # Azul estável
    nbins=40,
)

# 4. Cria o histograma de Fretes diretamente dos dados brutos
graph_freight = px.histogram(
    tabela_filtrada_frete,
    x="freight_value",
    title="<b>Distribuição dos Valores de Frete (Até R$ 100)</b>",
    labels={"freight_value": "Valor do Frete (R$)"},
    color_discrete_sequence=["#2ca02c"],  # Verde estável
    nbins=30,
)

# 5. Ajustes finos de layout para ambos os gráficos
for graph in [graph_price, graph_freight]:
    graph.update_layout(
        title_x=0.5,  # Centraliza o título
        yaxis_title="Total de Pedidos",
        bargap=0.05,  # Separação visual entre as barras
    )

# 6. Exibe os gráficos
graph_price.show()
graph_freight.show()
```

```python
import pandas as pd
import plotly.express as px

# 1. Carrega os dados otimizados
tabela = pd.read_csv(
    "olist_order_payments_dataset.csv",
    usecols=["payment_type", "payment_value"],
)

# 2. Calcula as proporções mantendo os valores numéricos decimais
contagem_payment = (
    tabela["payment_type"]
    .value_counts(normalize=True)
    .reset_index(name="Proporcao")
)

# 3. Cria o gráfico de barras interativo
graph_paymenttype = px.bar(
    contagem_payment,
    x="payment_type",
    y="Proporcao",
    title="<b>Distribuição de Tipos de Pagamento</b>",
    labels={
        "payment_type": "Forma de Pagamento",
        "Proporcao": "Percentual do Total",
    },
    text_auto=".1%",  # Mostra a porcentagem formatada certinho no topo da barra
    color="Proporcao",  # Aplica cores gradientes baseadas no volume
    color_continuous_scale="Blues",
)

# 4. Ajustes estéticos finais no layout
graph_paymenttype.update_layout(
    title_x=0.5,  # Centraliza o título
    showlegend=False,  # Oculta a legenda lateral redundante
    yaxis_tickformat=".0%",  # Formata o eixo Y como porcentagem inteira
)

# 5. Exibe o gráfico
graph_paymenttype.show()

```

```python
import pandas as pd
import plotly.express as px

# 1. Carrega os dados otimizados
tabela = pd.read_csv(
    "olist_products_dataset.csv",
    usecols=["product_category_name", "product_photos_qty"],
)

# 2. Calcula as proporções mantendo os valores numéricos decimais
contagem_category = (
    tabela["product_category_name"]
    .value_counts(normalize=True)
    .reset_index(name="Proporcao")
)
contagem_photos = (
    tabela["product_photos_qty"]
    .value_counts(normalize=True)
    .reset_index(name="Proporcao")
)

# 3. Gráfico 1: Top 15 Categorias de Produtos (Limitado para melhor visualização)
graph_category = px.bar(
    contagem_category.head(15),  # Mostra apenas as 15 maiores categorias
    x="product_category_name",
    y="Proporcao",
    title="<b>Top 15 Categorias de Produtos (% do Total)</b>",
    labels={
        "product_category_name": "Categoria",
        "Proporcao": "Percentual de Produtos",
    },
    text_auto=".1%",
    color="Proporcao",
    color_continuous_scale="Viridis",
)

graph_category.update_layout(
    title_x=0.5, xaxis_tickangle=-45, showlegend=False, yaxis_tickformat=".0%"
)

# 4. Gráfico 2: Quantidade de Fotos por Produto
# Convertemos a coluna para texto apenas no gráfico para tratar a quantidade de fotos como blocos individuais
graph_photos = px.bar(
    contagem_photos,
    x=contagem_photos["product_photos_qty"].astype(str),
    y="Proporcao",
    title="<b>Proporção de Fotos por Anúncio de Produto</b>",
    labels={
        "x": "Quantidade de Fotos",
        "Proporcao": "Percentual de Anúncios",
    },
    text_auto=".1%",
    color="Proporcao",
    color_continuous_scale="Plasma",
)

graph_photos.update_layout(title_x=0.5, showlegend=False, yaxis_tickformat=".0%")

# 5. Exibe os gráficos
graph_category.show()
graph_photos.show()
```

```python
import pandas as pd
import plotly.express as px

# --- 1. DATASET DE CLIENTES (Cidades e Estados) ---
tabela_clientes = pd.read_csv(
    "olist_customers_dataset.csv",
    usecols=["customer_city", "customer_state"],
)

# Calcula as proporções mantendo os valores numéricos decimais
contagem_cidade = (
    tabela_clientes["customer_city"]
    .value_counts(normalize=True)
    .reset_index(name="Proporcao")
)
contagem_estado = (
    tabela_clientes["customer_state"]
    .value_counts(normalize=True)
    .reset_index(name="Proporcao")
)

# --- 2. DATASET DE PAGAMENTOS ---
tabela_pagamentos = pd.read_csv(
    "olist_order_payments_dataset.csv", usecols=["payment_type"]
)

# Mantemos sem o normalize=True para contagem bruta
contagem_paymenttype = (
    tabela_pagamentos["payment_type"]
    .value_counts()
    .reset_index(name="Quantidade")
)


# --- 3. CRIAÇÃO DOS GRÁFICOS INTERATIVOS (px.bar) ---

# Gráfico 1: Tipos de Pagamento (Contagem Absoluta)
graph_paymenttype = px.bar(
    contagem_paymenttype,
    x="payment_type",
    y="Quantidade",
    title="<b>Contagem de Tipos de Pagamento</b>",
    labels={
        "payment_type": "Forma de Pagamento",
        "Quantidade": "Total de Pedidos",
    },
    text_auto=True,
    color="Quantidade",
    color_continuous_scale="Blues",
)
graph_paymenttype.update_layout(title_x=0.5, showlegend=False)

# Gráfico 2: Top 15 Cidades com mais Clientes (% do Total)
# CORRIGIDO: mudado de 'contagem_city' para 'contagem_cidade'
graph_city = px.bar(
    contagem_cidade.head(15),
    x="customer_city",
    y="Proporcao",
    title="<b>Top 15 Cidades com mais Clientes (% do Total)</b>",
    labels={
        "customer_city": "Cidade",
        "Proporcao": "Percentual de Clientes",
    },
    text_auto=".1%",
    color="Proporcao",
    color_continuous_scale="Viridis",
)
graph_city.update_layout(
    title_x=0.5, xaxis_tickangle=-45, showlegend=False, yaxis_tickformat=".0%"
)

# Gráfico 3: Clientes por Estado (% do Total)
graph_state = px.bar(
    contagem_estado,
    x="customer_state",
    y="Proporcao",
    title="<b>Distribuição de Clientes por Estado (% do Total)</b>",
    labels={
        "customer_state": "Estado",
        "Proporcao": "Percentual de Clientes",
    },
    text_auto=".1%",
    color="Proporcao",
    color_continuous_scale="Plasma",
)
graph_state.update_layout(
    title_x=0.5, xaxis_tickangle=-45, showlegend=False, yaxis_tickformat=".0%"
)


# --- 4. EXIBIÇÃO DOS GRÁFICOS ---
graph_paymenttype.show()
graph_city.show()
graph_state.show()
```
